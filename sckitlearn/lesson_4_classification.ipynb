```python
import pandas as pd 
from sklearn.tree import DecisionTreeRegressor
from sklearn.linear_model import LinearRegression
import kagglehub
from sklearn.model_selection import train_test_split
import os

path = kagglehub.dataset_download("yasserh/titanic-dataset")

csv_path = os.path.join(path, "Titanic-Dataset.csv")

titanic_data = pd.read_csv(csv_path)

print(titanic_data.head())
```

```python
print(titanic_data.head())
```

```python
print(titanic_data.columns)
```

```python
label = titanic_data["Survived"]
training_features = ["Name","Pclass","Sex","Age","Fare"]
```

```python
neccesary_features = titanic_data[training_features]
```

this time instead of deleting none value with drop na I will fill them with fillna
other then u know neural networks with numbers so sex male and female doesnt mean anything for it lets turn 1 male amd 0 female

```python
neccesary_features = neccesary_features.set_index("Name") 
neccesary_features['Age'] = neccesary_features['Age'].fillna(neccesary_features['Age'].mean())
neccesary_features['Sex'] = neccesary_features['Sex'].map({'male': 1, 'female': 0})
neccesary_features['Pclass'] = neccesary_features.Pclass.fillna(neccesary_features.Pclass.max())
train_X, val_X, train_y, val_y = train_test_split(neccesary_features, label, random_state=1, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier

titanic_model = DecisionTreeClassifier(random_state=1,
                                       max_depth = 5,
                                       min_samples_leaf = 3)
train_X.isnull().sum()
```

```python
titanic_model.fit(train_X, train_y)
titanic_predictions = titanic_model.predict(val_X)
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(val_y, titanic_predictions)

print("Confusion Matrix:")
print(cm)
```

classification report gives us precision, recall, f1-score, accuracy

```python
from sklearn.metrics import classification_report

print("Classification Report:")


print(classification_report(val_y, titanic_predictions))
```
